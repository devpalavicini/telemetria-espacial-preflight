```python
import random
testfalhou= False
relatorio = "Relatório da nave \n"

temp_intrn = random.uniform (15, 30) #temperatura interna
temp_ext = random.uniform (10, 35) #temperatura externa
intgrd_estr = random.randint (0, 1) #integridade estrutural

# Variáveis para definir o valor baseado em qual porcentagem de chance de cair True ou False
modulo = [True, False]
pesos = [0.80, 0.20]

#Status dos modulos 1, 2 e 3 usando a variável pesos
sta_mod1 = random.choices(modulo,weights = pesos,k=1)[0]
sta_mod2 = random.choices(modulo,weights = pesos,k=1)[0]
sta_mod3 = random.choices(modulo,weights = pesos,k=1)[0]

cap_total = random.uniform(500, 1000) #Capacidade total
carg_atual = random.randint(10, 100) #Carga atual %
cons_est_deco = random.uniform(10, 100) #Consumo estimado para decolagem
ener_utili = random.uniform(60.0, 85.0) #Variável para energia util
ener_consu = random.uniform(60.0, min(85.0, ener_consu)) #Variável para energia consumida

n = (ener_utili / ener_consu) * 100 #Calculo de rendimento
energia_disp = cap_total * (carg_atual / 100) #Calculo de energia disponivel
energia_util = energia_disp * (n / 100) #Calculo de energia util
auto_rest = energia_util - cons_est_deco #Calculo de autonomia restante
niv_enrg = energia_util

press_tanque = random.uniform(140, 210) #Variável de pressão dos tanques


if temp_intrn >= 18 and temp_intrn <= 25:
    relatorio += "Passou no verificação de temperatura interna. \n"

else:
    testfalhou = True

    relatorio += f"Falhou na verificação de temperatura interna, a temperatura está em {temp_intrn:.2f}. Valor esperado: entre 18C° e 25C° \n"

if temp_ext >= 20 and temp_ext <= 30:
    relatorio += "Passou na verificação da temperatura externa. \n"

else:
    testfalhou = True

    relatorio += f"Falhou na verificação de temperatura externa, a temperatura está em {temp_ext:.2f}. Valor esperado: entre 20C° e 30C°. \n"

if intgrd_estr == 1:
    relatorio += "Integridade estrutural dentro dos parâmetros esperados. \n"

else:
    testfalhou = True

    relatorio += f"Falhou no verificação de integridade estrutural: {intgrd_estr}. \n"

if auto_rest >= 0:
    relatorio += f"Energia suficiente para a decolagem, {niv_enrg:.2f}KWh restando {auto_rest:.2f}KWh após a decolagem. \n"

else:
    relatorio += f"Niveis de energia abaixo do esperado. {niv_enrg:.2f}KWh \n"
    testfalhou = True

if press_tanque >= 150 and press_tanque <= 200:
    relatorio += f"Pressão dos tanques de acordo com os parâmetros, a pressão está em {press_tanque:.2f}bar. \n"

elif press_tanque < 150:
    relatorio += f"Pressão dos tanques abaixo do esperado, a pressão é de {press_tanque:.2f}bar. \n "
    testfalhou = True

elif press_tanque > 200:
    relatorio += f"Pressão dos tanques acima do esperado, risco de explosão, a pressão é de {press_tanque:.2f}bar. \n"
    testfalhou = True

if sta_mod1  and sta_mod2  and sta_mod3:
    relatorio += "Status dos módulos: Todos os módulos passaram na verificação. \n"

else:
    relatorio += f"Status do módulo 1: {sta_mod1} \nStatus do módulo 2: {sta_mod2} \nStatus do módulo 3: {sta_mod3} \n"
    testfalhou = True

if testfalhou:
    print(relatorio)
    print("DECOLAGEM ABORTADA")

else:
    print("Todos os testes passaram")
    print(relatorio)
    print("PRONTO PARA DECOLAR")
```
